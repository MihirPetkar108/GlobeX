# Trade Risk Ensemble - Rebuild
Analyst scratchpad. IsolationForest + GRU concept; artifacts saved to rebuild.

```
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import RobustScaler
import joblib, json, torch
df=pd.read_csv('backend/brain/data/final_csv/04_trade_risk_eda.csv', nrows=8000)
print('Shape', df.shape)
num=[c for c in df.columns if df[c].dtype in ['float64','int64']][:10]
X=df[num].fillna(0).values
sc= RobustScaler(); Xs=sc.fit_transform(X)
clf=IsolationForest(n_estimators=100, contamination=0.05, random_state=42, n_jobs=-1)
clf.fit(Xs)
print('IF fitted.')
```
