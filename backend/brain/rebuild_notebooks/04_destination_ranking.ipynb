# Destination Ranking - MCDA Rebuild
Analyst scratchpad. Percentile-normalized multi-criteria engine; no supervised target added.

```
import pandas as pd, numpy as np, matplotlib.pyplot as plt
df=pd.read_csv('backend/brain/processed/destination_country_ranking_features.csv', nrows=1000)
print('Shape', df.shape); print('Cols', list(df.columns)[:6])
```

## Percentile-normalized score

```
score_cols=[c for c in df.columns if any(k in c.lower() for k in ['demand','growth','access']) and df[c].dtype in ['float64','int64']][:3]
if score_cols:
    for c in score_cols: df[c+'_pct']=df[c].rank(pct=True)
    df['ranking_score']=df[[c+'_pct' for c in score_cols]].mean(axis=1)
    top=df.sort_values('ranking_score',ascending=False).head(5)
    print(top[['importer_name','ranking_score']].head())
print('Ranking rebuilt.')
```
